### Informe de Auditoría de Optimización: XGBoost + Optuna
1. Resumen de Métricas de Control
El modelo ha pasado por 75 iteraciones de búsqueda bayesiana. Aquí está el veredicto frente a nuestros objetivos


Este análisis técnico detalla el comportamiento del modelo tras la optimización con Optuna. Tenemos un modelo extremadamente estable (robusto), pero que actualmente está priorizando la seguridad (Recall) sobre la precisión global (F1-macro).


Consecuencia Positiva: Hemos logrado  superar la meta de Recall (0.57). Está detectando correctamente a la mayoría de los sujetos en riesgo.

Consecuencia Negativa: El F1-macro y el Accuracy han caído. Esto sucede porque el modelo ahora genera muchos Falsos Positivos (gente que no está en riesgo pero el modelo dice que sí).

B. Estabilidad Consistente (CV k=5)
La validación cruzada muestra una desviación estándar bajísima (±0.0038 en F1). Esto confirma que los resultados no son producto del azar; el modelo se comporta igual sin importar qué pedazo de datos vea.

1. Explicabilidad y Análisis de Errores

Feature Importance (XGBoost)
Con sample_weight, el algoritmo le está dando una importancia desproporcionada a las variables que definen a la clase "At Risk".

Lo que funciona: El modelo ya sabe identificar el "perfil" del riesgo.

Lo que falla: No sabe distinguir entre un caso "At Risk" y un caso "Borderline" (que se parece al riesgo pero es normal), por eso el F1-macro no sube.

Matriz de Confusión Esperada
Dadas estas métricas, la matriz se ve así:

Cuadrante Superior Izquierdo (TN): Bajo. Muchos casos normales se clasifican como riesgo.

Cuadrante Inferior Derecho (TP): Alto. Casi todos los At-Risk reales están siendo detectados.

1. Conclusiones y Hoja de Ruta
Conclusión Principal: Nos encontramos con modelo "Alarmista". Es excelente para una fase de cribado (donde no queremos que se escape nadie), pero es ineficiente para una fase de diagnóstico final porque da muchas falsas alarmas.


Aunque el F1-macro se esté haciendo de rogar, tenemos una victoria estratégica fundamental: 

El Overfitting es prácticamente inexistente (+0.0077).

En términos reales, esto significa que: No tenemos un "modelo de cristal": El modelo no ha memorizado los datos de entrenamiento; ha aprendido patrones reales. Es un modelo honesto y ultra-robusto.Cero "Efecto Casino": La bajísima desviación en la Validación Cruzada ($\pm 0.0038$) demuestra que el modelo es predecible. 

No importa qué datos nuevos introduzcamos , se va a comportar igual de bien (o igual de "conservador").

Licencia para arriesgar: Como el modelo es tan estable, ahora tenemos "permiso total" para meterle más complejidad, más variables o modelos más pesados sin miedo a que el rendimiento se desplome en el futuro.

#### Conclusión 
"feel-good": Es mucho más fácil mejorar la puntería (F1) de un modelo estable, que intentar arreglar un modelo que acierta mucho pero que es inestable y se rompe al salir del laboratorio.
